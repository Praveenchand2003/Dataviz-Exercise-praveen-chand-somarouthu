```python
import pandas as pd
import numpy as np

df = pd.read_csv('/kaggle/input/datasets/ajaypalsinghlo/world-happiness-report-2023/WHR2023.csv')
print("\nColumn names:")
print(df.columns.tolist())

print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
print(df.head())
```

```python
import plotly.express as px
import plotly.graph_objects as go

print("Countries in dataset:")
print(df['Country name'].head())

print("\nScore range:",
      df['Ladder score'].min(),
      "–",
      df['Ladder score'].max())

print("\nBottom 10 countries:")
print(
    df.nsmallest(10, 'Ladder score')
      [['Country name', 'Ladder score']]
)
```

```python
#TASK 1
import plotly.express as px
import plotly.graph_objects as go

country_scores = (
df[['Country name', 'Ladder score']]
.sort_values('Ladder score', ascending=False)
)

fig = px.bar(
country_scores,
x='Ladder score',
y='Country name',
orientation='h',
text='Ladder score',
color='Country name',
title='Happiness Scores Across Countries',
labels={
'Ladder score': 'Happiness Score',
'Country name': 'Country'
}
)

fig.update_traces(
texttemplate='%{text:.2f}',
textposition='outside'
)

fig.update_layout(
xaxis=dict(
rangemode='tozero',
zeroline=True,
zerolinewidth=2
),
yaxis=dict(
categoryorder='total ascending'
),
template='plotly_white',
height=1200,
showlegend=False
)

fig.show()
```

```python
#TASK 2
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go


top_8 = df.nlargest(8, 'Ladder score').copy()


bottom_8 = df.nsmallest(8, 'Ladder score').copy()


top_8['Group'] = 'Happiest 8'
bottom_8['Group'] = 'Least Happy 8'


contrast_df = pd.concat([top_8, bottom_8])


country_codes = {
    'Finland': 'FIN',
    'Denmark': 'DEN',
    'Iceland': 'ISL',
    'Sweden': 'SWE',
    'Israel': 'ISR',
    'Netherlands': 'NED',
    'Norway': 'NOR',
    'Switzerland': 'SUI',
    'Afghanistan': 'AFG',
    'Lebanon': 'LBN',
    'Sierra Leone': 'SLE',
    'Zimbabwe': 'ZIM',
    'Congo (Kinshasa)': 'COD',
    'Botswana*': 'Botswana',
    'Malawi': 'MWI',
    'Comoros': 'Comoros'
}

contrast_df['Short Country'] = (
    contrast_df['Country name'].replace(country_codes)
)


global_average = df['Ladder score'].mean()


colors = [
    '#006400', '#228B22', '#2E8B57', '#3CB371',
    '#66CDAA', '#7CFC00', '#90EE90', '#98FB98',
    '#FFB6B6', '#FF9999', '#FF7F7F', '#FF6666',
    '#FF4D4D', '#FF3333', '#E63939', '#CC0000'
]


color_map = dict(
    zip(contrast_df['Short Country'], colors)
)


fig = px.bar(
    contrast_df,
    x='Short Country',
    y='Ladder score',
    color='Short Country',
    text='Ladder score',
    color_discrete_map=color_map,
    title='The Happiness Gap: Happiest Countries to Least Happy Countries',
    labels={
        'Short Country': 'Country',
        'Ladder score': 'Happiness Score'
    }
)


fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.add_hline(
    y=global_average,
    line_dash='dash',
    line_width=2,
    line_color='black',
    annotation_text=f'Global Average: {global_average:.2f}',
    annotation_position='top left'
)


fig.update_xaxes(
    tickangle=0
)

# Layout
fig.update_layout(
    template='plotly_white',
    height=700,
    showlegend=False,
    yaxis=dict(
        rangemode='tozero',
        zeroline=True,
        zerolinewidth=2
    )
)

fig.show()
```

```python
#Stretch Goal
import pandas as pd
import plotly.express as px


region_countries = {
    'Western Europe': [
        'Austria', 'Belgium', 'France', 'Germany', 'Ireland',
        'Luxembourg*', 'Netherlands', 'Switzerland', 'United Kingdom'
    ],

    'Latin America': [
        'Argentina', 'Bolivia', 'Brazil', 'Chile', 'Colombia',
        'Costa Rica', 'Cuba', 'Dominican Republic', 'Ecuador',
        'El Salvador', 'Guatemala', 'Honduras', 'Jamaica',
        'Mexico', 'Nicaragua', 'Panama', 'Paraguay', 'Peru',
        'Uruguay', 'Venezuela'
    ],

    'East Asia': [
        'China', 'Japan', 'South Korea', 'Mongolia'
    ],

    'Sub-Saharan Africa': [
        'Botswana*', 'Burkina Faso', 'Burundi', 'Cameroon',
        'Chad', 'Comoros', 'Congo (Brazzaville)',
        'Congo (Kinshasa)', 'Ethiopia', 'Ghana', 'Guinea',
        'Kenya', 'Liberia', 'Madagascar', 'Malawi', 'Mali',
        'Mauritania', 'Mauritius', 'Mozambique', 'Namibia',
        'Niger', 'Nigeria', 'Rwanda', 'Senegal',
        'Sierra Leone', 'Somalia', 'South Africa',
        'South Sudan', 'Tanzania', 'Togo', 'Uganda',
        'Zambia', 'Zimbabwe'
    ],

    'South Asia': [
        'Afghanistan', 'Bangladesh', 'Bhutan', 'India',
        'Maldives', 'Nepal', 'Pakistan', 'Sri Lanka'
    ]
}


df['Region'] = df['Country name'].map(
    {country: region for region, countries in region_countries.items()
     for country in countries}
)


region_df = df[df['Region'].notna()].copy()


region_summary = (
    region_df
    .groupby('Region')[
        ['Logged GDP per capita',
         'Freedom to make life choices']
    ]
    .mean()
    .reset_index()
)


chart_df = region_summary.melt(
    id_vars='Region',
    value_vars=[
        'Logged GDP per capita',
        'Freedom to make life choices'
    ],
    var_name='Sub-Factor',
    value_name='Average Score'
)


fig = px.bar(
    chart_df,
    x='Region',
    y='Average Score',
    color='Sub-Factor',
    barmode='group',
    title='Western Europe Leads in GDP and Freedom Compared with Other Regions',
    labels={
        'Region': 'Region',
        'Average Score': 'Average Score',
        'Sub-Factor': 'Sub-Factor'
    },
    color_discrete_map={
        'Logged GDP per capita': 'steelblue',
        'Freedom to make life choices': 'orange'
    }
)

fig.update_layout(
    template='plotly_white',
    height=600
)

fig.show()
```
